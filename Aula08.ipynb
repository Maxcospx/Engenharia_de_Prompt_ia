```python
import random
import time

def verificar_temperatura():
    # Em um sistema real, você obteria a temperatura de um sensor ou arquivo do sistema.
    # Por exemplo, em Linux, você pode ler de /sys/class/thermal/thermal_zone0/temp
    # ou usar uma biblioteca como psutil. Aqui, vamos simular.
    temperatura_atual = random.uniform(25.0, 35.0) # Simula uma temperatura entre 25 e 35 graus
    print("Temperatura atual do sistema: " + str(round(temperatura_atual, 2)) + "°C")

    if temperatura_atual > 30:
        print("ALERTA: Temperatura acima do limite! (" + str(round(temperatura_atual, 2)) + "°C)")
    else:
        print("Temperatura normal.")

# Chamando a função para verificar a temperatura
# verificar_temperatura() # Comentei esta linha pois o loop abaixo fará a verificação contínua.

# Você pode descomentar e ajustar o loop para verificar a temperatura periodicamente
while True:
    verificar_temperatura()
    time.sleep(5) # Verifica a cada 5 segundos
```
